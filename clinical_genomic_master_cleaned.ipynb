```python
master_df=pd.read_csv('cleaned_genomic_master.csv')
print(master_df.shape)
```

```python
missing_count= master_df.isnull().sum().sum()
print(f"Total missing values in master dataset:{missing_count}")
```

```python
null_cols=master_df.columns[master_df.isnull().any()]
print("Columns with missimg value and their counts:")
print(master_df[null_cols].isnull().sum())
```

```python
cols_to_fill=['PR.Status','ER.Status','HER2.Final.Status']
master_df[cols_to_fill]=master_df[cols_to_fill].fillna(-1)
print("Missing values after cleaning:")
print(master_df[cols_to_fill].isnull().sum())
```
